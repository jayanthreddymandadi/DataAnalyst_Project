```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- IMPORTANT ---
# Replace 'path/to/your/file.csv' with the actual file path where you saved the dataset.
# Example: 'C:/My-Internship-Projects/HR-Project/WA_Fn-UseC_-HR-Employee-Attrition.csv'

# Just add the 'r' at the beginning
file_path = r'C:\Users\manda\OneDrive\Documents\hr data.csv'
df = pd.read_csv(file_path)

# Display the first 5 rows to see what the data looks like
print("Data Head:")
print(df.head())
```

```python
# See all column names, their data types, and check for missing values
print("\nData Info:")
df.info()

# Get a statistical summary of all numerical columns
print("\nData Description:")
print(df.describe())

# Check how many employees have left ('Yes') vs. stayed ('No')
print("\nAttrition Counts:")
print(df['Attrition'].value_counts())
```

```python
# Drop columns that are not useful for prediction.
# 'EmployeeCount', 'StandardHours', 'Over18' have the same value for all employees.
# 'EmployeeNumber' is just an ID.
df_cleaned = df.drop(['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'], axis=1)

# Convert the 'Attrition' column to numerical format (1 for 'Yes', 0 for 'No')
# This is our target variable (what we want to predict)
df_cleaned['Attrition'] = df_cleaned['Attrition'].map({'Yes': 1, 'No': 0})

# Check the first 5 rows of our newly cleaned data
print("\nCleaned Data Head:")
print(df_cleaned.head())
```

```python
# --- FIX 2: Pointing to the OneDrive Desktop ---

# We add 'OneDrive' to the path
desktop_path = r'C:\Users\manda\OneDrive\Desktop\cleaned_hr_data.csv'

# Save the file to that specific path
df_cleaned.to_csv(desktop_path, index=False)

print(f"Successfully saved file to your OneDrive Desktop: {desktop_path}")
```

```python
#Step 5: Prepare Data for Modeling
# We'll use our 'df_cleaned' DataFrame from Phase 2
# pd.get_dummies() automatically finds all text columns and converts them
# drop_first=True is important to prevent a common modeling error
df_model = pd.get_dummies(df_cleaned, drop_first=True)

print("\nData after creating dummy variables:")
print(df_model.head())
```

```python
#Step 6: Define Features (X) and Target (y)
# Import the function for splitting our data
from sklearn.model_selection import train_test_split

# Define our features (X) and target (y)
X = df_model.drop('Attrition', axis=1)  # X is everything EXCEPT the 'Attrition' column
y = df_model['Attrition']              # y is ONLY the 'Attrition' column
```

```python
#Step 7: Split Data into Training and Testing Sets
# Split the data: 75% for training, 25% for testing
# random_state=42 ensures you get the same 'random' split I do, making your results reproducible
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print(f"\nTraining data shape: {X_train.shape}")
print(f"Testing data shape: {X_test.shape}")
```

```python
#Step 8: Train the Logistic Regression Model
# Import the model
from sklearn.linear_model import LogisticRegression

# 1. Create an instance of the model
# max_iter=1000 helps the model find the best solution without errors
model = LogisticRegression(max_iter=1000)

# 2. Train (fit) the model using our training data
model.fit(X_train, y_train)

print("\nModel trained successfully!")
```

```python
#Step 9: Evaluate the Model (Check Our Results)
# Import the evaluation metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 1. Make predictions using the test data
y_pred = model.predict(X_test)

# 2. Check the accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\n--- Model Evaluation ---")
print(f"Accuracy: {accuracy * 100:.2f}%")

# 3. Get the Confusion Matrix (a required deliverable)
# This shows us exactly where the model was right and wrong
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 4. Get the Classification Report (a required deliverable)
# This shows precision and recall for both '0' (Stayed) and '1' (Left)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
